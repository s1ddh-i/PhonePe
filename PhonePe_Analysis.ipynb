# PhonePe Transaction Insights
Analysis of digital transactions using Python

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("phonepe_transactions.csv")
df.head()
```

```python
df.shape
df.info()
df.describe()
```

```python
top_states = df.groupby("State")["Amount"].sum().sort_values(ascending=False).head(10)
top_states
top_states.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 States")
plt.show()
```

```python
types = df.groupby("Transaction_Type")["Amount"].sum()
types.plot(kind="pie", autopct="%1.1f%%")
plt.show()
```

```python
yearly = df.groupby("Year")["Amount"].sum()
yearly.plot(marker="o")
plt.title("Yearly Growth")
plt.show()
```

## Insights

- Telangana recorded highest transaction value.
- Peer-to-peer payments dominate usage.
- Strong yearly growth in digital payments.
- Merchant transactions are second highest.
